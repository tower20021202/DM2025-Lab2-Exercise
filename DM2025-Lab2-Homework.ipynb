## Competition Report

### 1. Preprocessing

- I merged three files by `id`:  
  `final_posts.json` (posts), `emotion.csv` (labels), `data_identification.csv` (train/test split).
- I cleaned the text with a `clean_text()` function:
  - removed URLs (`http...`, `https...`, `www...`)
  - removed placeholders like `[NAME]`, `[RELIGION]`
  - normalized spaces and lowercased all text
- I encoded the emotion labels using `LabelEncoder`:
  - `train_df["labels"] = label_encoder.fit_transform(train_df["emotion"])`
- I created a stratified train/validation split (`test_size=0.1`, `stratify=labels`)  
  to keep label distribution similar.
- I converted the data to Hugging Face `Dataset` objects for easy use with `Trainer`.

---

### 2. Feature Engineering

- Instead of manual features (e.g., TF–IDF), I relied on a pre-trained transformer model  
  to learn features automatically from text.
- I used the tokenizer of  
  `"j-hartmann/emotion-english-distilroberta-base"` with:
  - `padding="max_length"`, `truncation=True`, `max_length=96`
- After tokenization, I kept only:
  - train/val: `input_ids`, `attention_mask`, `labels`
  - test: `input_ids`, `attention_mask`
- I called `set_format("torch")` so everything becomes PyTorch tensors.

---

### 3. Model Explanation

- I used `AutoModelForSequenceClassification` with the checkpoint  
  `"j-hartmann/emotion-english-distilroberta-base"`.
- The original model has 7 classes; the competition has 6, so I set:
  - `num_labels = NUM_LABELS`
  - `ignore_mismatched_sizes = True`  
  → this keeps the pre-trained encoder and replaces the classification head with a new 6-class head.
- I trained with Hugging Face `Trainer`:
  - metric: accuracy (`compute_metrics` uses `accuracy_score`)
  - batch sizes: 32 (train) / 64 (eval)
  - learning rate: `3e-5`, weight decay: `0.01`, warmup steps: `500`
  - epochs: 5 with GPU; 1 on CPU.
- I added `EarlyStoppingCallback` (patience=3) to stop when validation loss stops improving.

---

### 4. Bonus: What I Tried and Insights

- **Model choice:**  
  Using an emotion-focused DistilRoBERTa checkpoint works better in this task than a generic model,  
  because it already “knows” a lot about emotional language.
- **Text cleaning:**  
  Removing URLs and placeholders made training a bit more stable and slightly improved validation performance.
- **Stratified split + early stopping:**  
  Stratified splitting gives a more reliable validation score,  
  and early stopping saves time and reduces overfitting.

# **Competition Code**

## 1. Preprocessing Steps

```python
# 階段一：資料前處理 (Preprocessing Steps)
# 涵蓋：檔案載入、資料合併、資料集劃分、文本清洗


import pandas as pd
import json
import os
import re
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder # 標籤編碼，也可視為特徵工程
from sklearn.metrics import accuracy_score
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback,
)
import torch
from time import time
import inspect
from datasets import Dataset

BASE_PATH = "/content/drive/MyDrive/DM kaggle/dm-lab-2-private-competition/"
SUBMISSION_PATH = "/content/drive/MyDrive/DM kaggle/dm-lab-2-private-competition/submission_sota_emotion.csv"

JSON_PATH = BASE_PATH + "final_posts.json"
EMOTION_PATH = BASE_PATH + "emotion.csv"
SPLIT_PATH = BASE_PATH + "data_identification.csv"

# SOTA 模型替換 
MODEL_CHECKPOINT = "j-hartmann/emotion-english-distilroberta-base"


print("--- 1. 資料載入、清洗與標籤準備 ---")
try:
    with open(JSON_PATH, "r") as f:
        raw_data = json.load(f)

    posts_data = [
        {
            "id": item["root"]["_source"]["post"]["post_id"],
            "text": item["root"]["_source"]["post"]["text"],
        }
        for item in raw_data
    ]
    df_posts = pd.DataFrame(posts_data)

    # 載入情緒標籤與資料集劃分資訊
    df_emotions = pd.read_csv(EMOTION_PATH)
    df_split = pd.read_csv(SPLIT_PATH)

    # 資料合併
    df_combined = (
        df_posts.merge(df_split, on="id", how="left")
                .merge(df_emotions, on="id", how="left")
    )

    # 依 'split' 欄位劃分訓練集和測試集（原始資料）
    train_df = (
        df_combined[df_combined["split"] == "train"]
        .copy()
        .reset_index(drop=True)
    )
    test_df_raw = (
        df_combined[df_combined["split"] == "test"]
        .copy()
        .reset_index(drop=True)
    )

except FileNotFoundError as e:
    print(f"致命錯誤：找不到檔案。錯誤: {e}")
    raise


def clean_text(text):
    if not isinstance(text, str):
        return ""
    # 清掉網址與 [NAME] / [RELIGION]
    text = re.sub(
        r"http\S+|www\S+|https\S+|\[NAME\]|\[RELIGION\]",
        "",
        text,
    )
    # 統一小寫並清理多餘空白
    text = re.sub(r"\s+", " ", text).strip().lower()
    return text

train_df["text"] = train_df["text"].apply(clean_text)
test_df_raw["text"] = test_df_raw["text"].apply(clean_text)

label_encoder = LabelEncoder()
train_df["labels"] = label_encoder.fit_transform(train_df["emotion"])
NUM_LABELS = len(label_encoder.classes_)
print("情緒類別：", label_encoder.classes_)

train_data, val_data = train_test_split(
    train_df,
    test_size=0.1,
    random_state=42,
    stratify=train_df["labels"],
)

train_hf_dataset = Dataset.from_pandas(
    train_data[["text", "labels", "id"]].reset_index(drop=True)
)
val_hf_dataset = Dataset.from_pandas(
    val_data[["text", "labels", "id"]].reset_index(drop=True)
)
test_hf_dataset = Dataset.from_pandas(
    test_df_raw[["text", "id"]].reset_index(drop=True)
)
```

## 2. Feature Engineering Steps


```python
# 階段二：特徵工程 (Feature Engineering Steps)
# 涵蓋：Tokenization（將文本轉換為模型可理解的輸入特徵）

print("\n--- 2. Tokenization ---")
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)


def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=96,
    )


tokenized_train_dataset = train_hf_dataset.map(tokenize_function, batched=True)
tokenized_val_dataset = val_hf_dataset.map(tokenize_function, batched=True)
tokenized_test_dataset = test_hf_dataset.map(tokenize_function, batched=True)

print("移除不需要的欄位 ('text', 'id')...")
tokenized_train_dataset = tokenized_train_dataset.remove_columns(["text", "id"])
tokenized_val_dataset = tokenized_val_dataset.remove_columns(["text", "id"])
tokenized_test_dataset = tokenized_test_dataset.remove_columns(["text", "id"])

tokenized_train_dataset.set_format("torch")
tokenized_val_dataset.set_format("torch")
tokenized_test_dataset.set_format("torch")
```

## 3. Model Implementation Steps

```python
# 階段三：模型實作 (Model Implementation Steps)
# 涵蓋：模型載入、訓練參數設定、訓練、預測與結果輸出
print("\n--- 3. BERT 模型 Fine-tuning ---")


model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_CHECKPOINT,
    num_labels=NUM_LABELS,
    ignore_mismatched_sizes=True, 
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return {"accuracy": accuracy_score(labels, predictions)}


use_cuda = torch.cuda.is_available()
if use_cuda:
    num_epochs = 5  
else:
    num_epochs = 1  

base_training_args = dict(
    output_dir="./results_distilroberta",
    num_train_epochs=num_epochs,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=64,
    warmup_steps=500,
    weight_decay=0.01,
    learning_rate=3e-5,
    logging_dir="./logs_distilroberta",
    logging_steps=200,
    evaluation_strategy="epoch",  
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    greater_is_better=True,
    seed=42,
    report_to="none",
    disable_tqdm=False,
)

sig = inspect.signature(TrainingArguments)
adapted_args = base_training_args.copy()


if "evaluation_strategy" not in sig.parameters:
    if "eval_strategy" in sig.parameters and "evaluation_strategy" in adapted_args:
        adapted_args["eval_strategy"] = adapted_args.pop("evaluation_strategy")
        print("偵測到使用 'eval_strategy' 參數命名，已自動轉換。")
    else:
        adapted_args.pop("evaluation_strategy", None)
        adapted_args["load_best_model_at_end"] = False
        print("當前版本不支援 evaluation_strategy，已移除且關閉 load_best_model_at_end。")

if "save_strategy" not in sig.parameters and "save_strategy" in adapted_args:
    adapted_args.pop("save_strategy", None)
    print("當前版本不支援 save_strategy，已自動移除。")

valid_kwargs = {k: v for k, v in adapted_args.items() if k in sig.parameters}
training_args = TrainingArguments(**valid_kwargs)

early_stopping = EarlyStoppingCallback(
    early_stopping_patience=3,
    early_stopping_threshold=0.0005,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_val_dataset,
    compute_metrics=compute_metrics,
    callbacks=[early_stopping],
)

start_time = time()
if use_cuda:
    print("偵測到 CUDA。開始 Fine-tuning（最多約 5 個 epoch，視 EarlyStopping 而定）。")
else:
    print("未偵測到 CUDA，在 CPU 上訓練，目前只設定 1 個 epoch。")

trainer.train()

end_time = time()
print(f"\n模型 Fine-tuning 完成。總耗時: {end_time - start_time:.2f} 秒")


print("\n--- 4. 測試集預測與提交檔案生成 ---")

predictions_output = trainer.predict(tokenized_test_dataset)
logits = predictions_output.predictions
predicted_labels_indices = np.argmax(logits, axis=1)

predicted_emotions = label_encoder.inverse_transform(predicted_labels_indices)

submission_df = test_df_raw[["id"]].copy()
submission_df["emotion"] = predicted_emotions
submission_df.columns = ["id", "emotion"]

os.makedirs(os.path.dirname(SUBMISSION_PATH), exist_ok=True)
submission_df.to_csv(SUBMISSION_PATH, index=False)

print("\n--- 競賽提交完成 ---")
print(f"生成的提交檔案已儲存至: {SUBMISSION_PATH}")
```
